```python
gdp_sorted = gdp.sort_values(["Entity", "Year"])
gdp_sorted["yoy_change"] = gdp_sorted.groupby("Entity")["gdp"].diff()
```

```python
def decline_then_increase(changes):
    seen_decline = False
    for c in changes.dropna():
        if c < 0:
            seen_decline = True
        if seen_decline and c > 0:
            return True
    return False

country_patterns = (
    gdp_sorted.groupby("Entity")["yoy_change"]
    .apply(decline_then_increase)
    .reset_index()
    .rename(columns={"yoy_change": "decline_then_increase"})
)
```

```python
total_gdp = gdp_sorted.groupby("Entity")["gdp"].sum().reset_index()
patterned_gdp = total_gdp.merge(country_patterns, on="Entity")

# Keep only decline-then-increase countries
recovered_countries = patterned_gdp[patterned_gdp["decline_then_increase"]]
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

top10_recovered = recovered_countries.nlargest(10, "gdp")

sns.barplot(
    data=top10_recovered,
    x="gdp",
    y="Entity",
    palette="Greens"
)
plt.title("Top 10 Countries: GDP Decline Followed by Recovery")
plt.xlabel("Total GDP (millions of US$)")
plt.ylabel("Country")
plt.show()
```

```python
gdp_sorted["gdp_quantile"] = pd.qcut(
    gdp_sorted["gdp"], 4, labels=["Q1 (Lowest)", "Q2", "Q3", "Q4 (Highest)"]
)

# Merge flags
gdp_with_patterns = gdp_sorted.merge(
    country_patterns.rename(columns={"decline_then_increase": "pattern_flag"}),
    on="Entity"
)

# Probability calculation
q1_and_recovered = gdp_with_patterns[
    (gdp_with_patterns["pattern_flag"]) & (gdp_with_patterns["gdp_quantile"] == "Q1 (Lowest)")
].shape[0]

recovered_total = gdp_with_patterns[gdp_with_patterns["pattern_flag"]].shape[0]

p_q1_given_recovered = q1_and_recovered / recovered_total if recovered_total > 0 else 0
print("Probability:", p_q1_given_recovered)
```

```python
import pandas as pd

# Sort the dataframe to ensure proper time order
gdp_sorted = gdp.sort_values(["Entity", "Year"])

# Compute year-over-year growth per country
gdp_sorted["gdp_diff"] = gdp_sorted.groupby("Entity")["gdp"].diff()

# Flag declines and increases
gdp_sorted["decline"] = gdp_sorted["gdp_diff"] < 0
gdp_sorted["increase"] = gdp_sorted["gdp_diff"] > 0

# Function to check comeback pattern
def has_comeback(group):
    declines = (group["decline"].rolling(2).sum() == 2)  # 2 years decline
    increases = (group["increase"].rolling(2).sum() == 2)  # 2 years increase
    return any(declines & increases.shift(-2))  # decline followed by later increase

# Apply per country
comeback_countries = (
    gdp_sorted.groupby("Entity")
    .apply(has_comeback)
    .reset_index(name="comeback")
)

# Filter only comeback countries
comeback_list = comeback_countries[comeback_countries["comeback"]]["Entity"].tolist()

print("Comeback Countries:", comeback_list)
```

```python
import pandas as pd

# Ensure sorted by country and year
gdp_sorted = gdp.sort_values(["Entity", "Year"]).copy()

# Calculate year-over-year difference
gdp_sorted["gdp_diff"] = gdp_sorted.groupby("Entity")["gdp"].diff()

# Flag direction
gdp_sorted["trend"] = gdp_sorted["gdp_diff"].apply(lambda x: "decline" if x < 0 else ("increase" if x > 0 else "flat"))

# Function to measure longest decline→increase run
def longest_comeback(group):
    trends = group["trend"].tolist()
    
    max_decline, max_increase = 0, 0
    current_decline, current_increase = 0, 0
    found_decline = False
    
    for t in trends:
        if t == "decline":
            current_decline += 1
            current_increase = 0
            found_decline = True
        elif t == "increase":
            if found_decline:  # only start counting if we had declines
                current_increase += 1
            else:
                current_increase = 0
            max_increase = max(max_increase, current_increase)
        else:  # flat
            current_decline = 0
            current_increase = 0
        
        max_decline = max(max_decline, current_decline)
    
    return pd.Series({"longest_decline": max_decline, "longest_increase": max_increase})

# Apply per country
comeback_strength = (
    gdp_sorted.groupby("Entity")
    .apply(longest_comeback)
    .reset_index()
)

# Rank by decline length + increase length
comeback_strength["score"] = comeback_strength["longest_decline"] + comeback_strength["longest_increase"]
top_comebacks = comeback_strength.sort_values("score", ascending=False).head(10)

print(top_comebacks)
```
