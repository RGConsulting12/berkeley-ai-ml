To sort the resulting values by medians from greatest to least, first flatten and rename the columns using the code:


````python
df = gapminder
````

````python
from sklearn.preprocessing import LabelEncoder

country_encoder = LabelEncoder()
continent_encoder = LabelEncoder()

df["country_id"] = country_encoder.fit_transform(df["country"])
df["continent_id"] = continent_encoder.fit_transform(df["continent"])
````

````python
vectors = df[["country_id", "continent_id", "year", "lifeExp", "pop", "gdpPercap"]].values

print("Shape of vectors:", vectors.shape)
print("Example vectors:\n", vectors[:5])
````

````python
import numpy as np
np.save("gapminder_vectors.npy", vectors)
````

````python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.lineplot(data=df, x="year", y="lifeExp", hue="continent")
plt.title("Life Expectancy Trends by Continent")
plt.ylabel("Life Expectancy")
plt.xlabel("Year")
plt.show()
````

````python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df[df["year"]==2007], 
    x="gdpPercap", y="lifeExp", 
    size="pop", hue="continent", alpha=0.6, sizes=(20, 2000)
)
plt.xscale("log")
plt.title("GDP per Capita vs Life Expectancy (2007)")
plt.show()
````

````python
corr = df[["lifeExp", "pop", "gdpPercap", "year"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
````

````python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Reduce to 2D
pca = PCA(n_components=2)
reduced = pca.fit_transform(vectors)

plt.figure(figsize=(10,6))
plt.scatter(reduced[:,0], reduced[:,1], alpha=0.6, c=df["continent_id"])
plt.title("Countries Vectorized & Reduced with PCA")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
````

````python
p = ( (df["continent"]=="Europe") & (df["lifeExp"]>70) ).sum() / (df["continent"]=="Europe").sum()
print("P(lifeExp > 70 | Europe) =", p)
````

````python
import plotly.express as px

fig = px.scatter_3d(
    df[df["year"]==2007],  # pick a year to simplify
    x="gdpPercap",
    y="lifeExp",
    z="pop",
    color="continent",
    size="pop",
    hover_name="country",
    log_x=True  # log scale for GDP
)

fig.update_layout(title="3D Scatter: GDP, Life Expectancy, Population (2007)")
fig.show()
````

````python
fig = px.line_3d(
    df[df["country"].isin(["India", "China", "United States"])],
    x="year",
    y="lifeExp",
    z="gdpPercap",
    color="country"
)

fig.update_layout(title="3D Line Plot: Development Trajectories")
fig.show()
````

````python
import pandas as pd

# Aggregate by continent & year
agg = df.groupby(["continent", "year"])["gdpPercap"].mean().reset_index()
pivot = agg.pivot(index="year", columns="continent", values="gdpPercap")

fig = px.surface(
    z=pivot.values,
    x=pivot.columns,   # continents
    y=pivot.index,     # years
    title="3D Surface: Avg GDP per Capita by Continent"
)
fig.show()
````
