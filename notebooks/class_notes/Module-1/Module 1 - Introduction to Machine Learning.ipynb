- \( X \): Input features

- \( \hat{y} \): Predicted output

```python
from sklearn.datasets import make_classification
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

X, y = make_classification(
    n_samples=200,       # number of rows
    n_features=4,        # total features
    n_informative=2,     # features that influence the target
    n_redundant=0,       # features that are combinations of others
    n_classes=2,         # binary classification
    random_state=42
)

df = pd.DataFrame(X, columns=["Feature_1", "Feature_2", "Feature_3", "Feature_4"])
df["Target"] = y
df.head()

sns.scatterplot(data=df, x="Feature_1", y="Feature_2", hue="Target", palette="coolwarm")
plt.title("Simulated Classification Data")
plt.show()
```

```python
# 3. Discrete vs Continuous Probability

from scipy.stats import binom, norm
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 11)
plt.bar(x, binom.pmf(x, 10, 0.5))
plt.title("Discrete Probability (Binomial)")
plt.show()

x = np.linspace(-3, 3, 100)
plt.plot(x, norm.pdf(x))
plt.title("Continuous Probability (Normal Distribution)")
plt.show()
```

```python
# Data Exploration and Basic Statistics
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from collections import Counter
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')

# Set style for better-looking plots
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

# Convert to pandas and explore
df = ds["train"].to_pandas()

print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
print("\nData Types:")
print(df.dtypes)
print("\nFirst few rows:")
df.head()
```

```python
# 1. Categorical Data Visualizations

# Create a figure with multiple subplots
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('Categorical Data Analysis', fontsize=16, fontweight='bold')

# 1. Primary Category Distribution
category_counts = df['primary_category'].value_counts()
axes[0, 0].pie(category_counts.values, labels=category_counts.index, autopct='%1.1f%%', startangle=90)
axes[0, 0].set_title('Distribution of Primary Categories')

# 2. Top Authors
top_authors = df['author'].value_counts().head(10)
axes[0, 1].barh(range(len(top_authors)), top_authors.values)
axes[0, 1].set_yticks(range(len(top_authors)))
axes[0, 1].set_yticklabels(top_authors.index)
axes[0, 1].set_title('Top 10 Authors by Number of Servers')
axes[0, 1].set_xlabel('Number of Servers')

# 3. Category vs Author Heatmap (Top 10 authors)
top_10_authors = df['author'].value_counts().head(10).index
heatmap_data = df[df['author'].isin(top_10_authors)].groupby(['primary_category', 'author']).size().unstack(fill_value=0)
sns.heatmap(heatmap_data, annot=True, fmt='d', cmap='YlOrRd', ax=axes[1, 0])
axes[1, 0].set_title('Category vs Author Heatmap (Top 10 Authors)')
axes[1, 0].set_xlabel('Author')
axes[1, 0].set_ylabel('Primary Category')

# 4. Server Count by Category (Bar Chart)
category_counts.plot(kind='bar', ax=axes[1, 1], color='skyblue', edgecolor='black')
axes[1, 1].set_title('Number of Servers by Category')
axes[1, 1].set_xlabel('Primary Category')
axes[1, 1].set_ylabel('Count')
axes[1, 1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# Print some statistics
print(f"Total number of servers: {len(df)}")
print(f"Number of unique categories: {df['primary_category'].nunique()}")
print(f"Number of unique authors: {df['author'].nunique()}")
print(f"Most common category: {df['primary_category'].mode()[0]}")
print(f"Most prolific author: {df['author'].mode()[0]}")
```

```python
# 2. Temporal Analysis and Trends

# Convert release_date to datetime
df['release_date'] = pd.to_datetime(df['release_date'])
df['year'] = df['release_date'].dt.year
df['month'] = df['release_date'].dt.month
df['quarter'] = df['release_date'].dt.quarter

# Create temporal visualizations
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Temporal Analysis of MCP Servers', fontsize=16, fontweight='bold')

# 1. Release Timeline
monthly_releases = df.groupby([df['release_date'].dt.to_period('M')]).size()
monthly_releases.plot(kind='line', ax=axes[0, 0], marker='o', linewidth=2, markersize=4)
axes[0, 0].set_title('Monthly Server Releases Over Time')
axes[0, 0].set_xlabel('Month')
axes[0, 0].set_ylabel('Number of Servers Released')
axes[0, 0].grid(True, alpha=0.3)

# 2. Releases by Year
yearly_releases = df['year'].value_counts().sort_index()
axes[0, 1].bar(yearly_releases.index, yearly_releases.values, color='lightcoral', edgecolor='black')
axes[0, 1].set_title('Server Releases by Year')
axes[0, 1].set_xlabel('Year')
axes[0, 1].set_ylabel('Number of Servers')
axes[0, 1].grid(True, alpha=0.3)

# 3. Category Trends Over Time (Top 5 categories)
top_categories = df['primary_category'].value_counts().head(5).index
for category in top_categories:
    category_data = df[df['primary_category'] == category]
    monthly_category = category_data.groupby([category_data['release_date'].dt.to_period('M')]).size()
    axes[1, 0].plot(monthly_category.index.astype(str), monthly_category.values, 
                    marker='o', label=category, linewidth=2, markersize=3)

axes[1, 0].set_title('Top 5 Categories: Monthly Release Trends')
axes[1, 0].set_xlabel('Month')
axes[1, 0].set_ylabel('Number of Servers')
axes[1, 0].legend(bbox_to_anchor=(1.05, 1), loc='upper left')
axes[1, 0].tick_params(axis='x', rotation=45)
axes[1, 0].grid(True, alpha=0.3)

# 4. Quarterly Distribution
quarterly_releases = df['quarter'].value_counts().sort_index()
quarters = ['Q1', 'Q2', 'Q3', 'Q4']
axes[1, 1].pie(quarterly_releases.values, labels=quarters, autopct='%1.1f%%', startangle=90)
axes[1, 1].set_title('Distribution of Releases by Quarter')

plt.tight_layout()
plt.show()

# Print temporal statistics
print("Temporal Statistics:")
print(f"Date range: {df['release_date'].min().strftime('%Y-%m-%d')} to {df['release_date'].max().strftime('%Y-%m-%d')}")
print(f"Most active year: {df['year'].mode()[0]} ({df['year'].value_counts().max()} releases)")
print(f"Most active quarter: Q{df['quarter'].mode()[0]} ({df['quarter'].value_counts().max()} releases)")
print(f"Average releases per month: {len(df) / df['release_date'].dt.to_period('M').nunique():.1f}")
```

```python
# 3. Text Analysis and Word Clouds

from wordcloud import WordCloud
from collections import Counter
import re

# Create text analysis visualizations
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Text Analysis of MCP Server Descriptions and Titles', fontsize=16, fontweight='bold')

# 1. Title Length Distribution
df['title_length'] = df['title'].str.len()
axes[0, 0].hist(df['title_length'], bins=20, color='lightblue', edgecolor='black', alpha=0.7)
axes[0, 0].set_title('Distribution of Title Lengths')
axes[0, 0].set_xlabel('Title Length (characters)')
axes[0, 0].set_ylabel('Frequency')
axes[0, 0].grid(True, alpha=0.3)

# 2. Description Length Distribution
df['description_length'] = df['description'].str.len()
axes[0, 1].hist(df['description_length'], bins=30, color='lightgreen', edgecolor='black', alpha=0.7)
axes[0, 1].set_title('Distribution of Description Lengths')
axes[0, 1].set_xlabel('Description Length (characters)')
axes[0, 1].set_ylabel('Frequency')
axes[0, 1].grid(True, alpha=0.3)

# 3. Most Common Words in Titles
def extract_words(text):
    if pd.isna(text):
        return []
    # Remove special characters and convert to lowercase
    words = re.findall(r'\b[a-zA-Z]+\b', str(text).lower())
    # Filter out common stop words
    stop_words = {'the', 'a', 'an', 'and', 'or', 'but', 'in', 'on', 'at', 'to', 'for', 'of', 'with', 'by', 'is', 'are', 'was', 'were', 'be', 'been', 'have', 'has', 'had', 'do', 'does', 'did', 'will', 'would', 'could', 'should', 'may', 'might', 'must', 'can', 'this', 'that', 'these', 'those', 'i', 'you', 'he', 'she', 'it', 'we', 'they', 'me', 'him', 'her', 'us', 'them'}
    return [word for word in words if word not in stop_words and len(word) > 2]

title_words = []
for title in df['title']:
    title_words.extend(extract_words(title))

word_counts = Counter(title_words)
top_words = dict(word_counts.most_common(15))

words = list(top_words.keys())
counts = list(top_words.values())
axes[1, 0].barh(range(len(words)), counts, color='orange', edgecolor='black')
axes[1, 0].set_yticks(range(len(words)))
axes[1, 0].set_yticklabels(words)
axes[1, 0].set_title('Top 15 Most Common Words in Titles')
axes[1, 0].set_xlabel('Frequency')

# 4. Word Cloud of Descriptions
description_text = ' '.join(df['description'].dropna().astype(str))
wordcloud = WordCloud(width=800, height=400, background_color='white', 
                     max_words=100, colormap='viridis').generate(description_text)
axes[1, 1].imshow(wordcloud, interpolation='bilinear')
axes[1, 1].axis('off')
axes[1, 1].set_title('Word Cloud of Server Descriptions')

plt.tight_layout()
plt.show()

# Print text statistics
print("Text Analysis Statistics:")
print(f"Average title length: {df['title_length'].mean():.1f} characters")
print(f"Average description length: {df['description_length'].mean():.1f} characters")
print(f"Longest title: {df['title_length'].max()} characters")
print(f"Longest description: {df['description_length'].max()} characters")
print(f"Most common word in titles: '{word_counts.most_common(1)[0][0]}' ({word_counts.most_common(1)[0][1]} times)")
```

```python
# 4. Correlation and Relationship Analysis

# Create numerical features for correlation analysis
df_numeric = df.copy()

# Convert categorical variables to numerical for correlation analysis
df_numeric['category_encoded'] = pd.Categorical(df_numeric['primary_category']).codes
df_numeric['author_encoded'] = pd.Categorical(df_numeric['author']).codes
df_numeric['year_numeric'] = df_numeric['year']
df_numeric['month_numeric'] = df_numeric['month']

# Create correlation matrix
correlation_cols = ['rank', 'page_num', 'page_idx', 'title_length', 'description_length', 
                   'category_encoded', 'author_encoded', 'year_numeric', 'month_numeric']
correlation_matrix = df_numeric[correlation_cols].corr()

# Create relationship visualizations
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Correlation and Relationship Analysis', fontsize=16, fontweight='bold')

# 1. Correlation Heatmap
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, 
            square=True, ax=axes[0, 0], cbar_kws={'shrink': 0.8})
axes[0, 0].set_title('Correlation Matrix of Numerical Features')

# 2. Rank vs Page Number Scatter Plot
axes[0, 1].scatter(df_numeric['page_num'], df_numeric['rank'], alpha=0.6, color='purple')
axes[0, 1].set_xlabel('Page Number')
axes[0, 1].set_ylabel('Rank')
axes[0, 1].set_title('Rank vs Page Number Relationship')
axes[0, 1].grid(True, alpha=0.3)

# Add trend line
z = np.polyfit(df_numeric['page_num'], df_numeric['rank'], 1)
p = np.poly1d(z)
axes[0, 1].plot(df_numeric['page_num'], p(df_numeric['page_num']), "r--", alpha=0.8)

# 3. Title Length vs Description Length
axes[1, 0].scatter(df_numeric['title_length'], df_numeric['description_length'], 
                  alpha=0.6, color='green')
axes[1, 0].set_xlabel('Title Length (characters)')
axes[1, 0].set_ylabel('Description Length (characters)')
axes[1, 0].set_title('Title Length vs Description Length')
axes[1, 0].grid(True, alpha=0.3)

# Add trend line
z = np.polyfit(df_numeric['title_length'], df_numeric['description_length'], 1)
p = np.poly1d(z)
axes[1, 0].plot(df_numeric['title_length'], p(df_numeric['title_length']), "r--", alpha=0.8)

# 4. Box Plot: Description Length by Category (Top 5 categories)
top_5_categories = df['primary_category'].value_counts().head(5).index
df_top_categories = df[df['primary_category'].isin(top_5_categories)]
sns.boxplot(data=df_top_categories, x='primary_category', y='description_length', ax=axes[1, 1])
axes[1, 1].set_title('Description Length Distribution by Category (Top 5)')
axes[1, 1].set_xlabel('Primary Category')
axes[1, 1].set_ylabel('Description Length (characters)')
axes[1, 1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# Print correlation insights
print("Correlation Analysis Insights:")
print(f"Strongest positive correlation: {correlation_matrix.unstack().sort_values(ascending=False).iloc[1]}")
print(f"Strongest negative correlation: {correlation_matrix.unstack().sort_values(ascending=True).iloc[0]}")
print(f"Rank-Page Number correlation: {correlation_matrix.loc['rank', 'page_num']:.3f}")
print(f"Title-Description length correlation: {correlation_matrix.loc['title_length', 'description_length']:.3f}")

# Additional insights
print(f"\nAdditional Insights:")
print(f"Average rank: {df['rank'].mean():.1f}")
print(f"Average page number: {df['page_num'].mean():.1f}")
print(f"Servers with highest ranks (most popular): {df.nsmallest(5, 'rank')[['title', 'rank']].to_string(index=False)}")
```

```python
# Growth Phase Analysis and Trend Detection

from scipy import stats
import numpy as np

print("🔍 GROWTH PHASE ANALYSIS")
print("=" * 50)

# Define growth phases based on cumulative growth
df_sorted = df.sort_values('release_date')
df_sorted['cumulative'] = range(1, len(df_sorted) + 1)
df_sorted['days_since_start'] = (df_sorted['release_date'] - df_sorted['release_date'].iloc[0]).dt.days

# Calculate growth rates for different periods
def calculate_growth_rate(data, period_days=30):
    """Calculate growth rate over specified period"""
    growth_rates = []
    for i in range(period_days, len(data)):
        start_idx = i - period_days
        start_count = data.iloc[start_idx]['cumulative']
        end_count = data.iloc[i]['cumulative']
        growth_rate = ((end_count - start_count) / start_count) * 100 if start_count > 0 else 0
        growth_rates.append(growth_rate)
    return growth_rates

# Calculate 30-day growth rates
growth_rates_30d = calculate_growth_rate(df_sorted, 30)
df_sorted['growth_rate_30d'] = [np.nan] * 30 + growth_rates_30d

# Identify growth phases
print("\n�� GROWTH PHASE IDENTIFICATION:")

# Early phase (first 25% of data)
early_cutoff = int(len(df_sorted) * 0.25)
early_data = df_sorted.iloc[:early_cutoff]
early_avg_growth = early_data['growth_rate_30d'].mean()

# Middle phase (middle 50% of data)
middle_start = int(len(df_sorted) * 0.25)
middle_end = int(len(df_sorted) * 0.75)
middle_data = df_sorted.iloc[middle_start:middle_end]
middle_avg_growth = middle_data['growth_rate_30d'].mean()

# Recent phase (last 25% of data)
recent_start = int(len(df_sorted) * 0.75)
recent_data = df_sorted.iloc[recent_start:]
recent_avg_growth = recent_data['growth_rate_30d'].mean()

print(f"   Early Phase (First 25%): {early_avg_growth:.2f}% avg growth rate")
print(f"   Middle Phase (Middle 50%): {middle_avg_growth:.2f}% avg growth rate")
print(f"   Recent Phase (Last 25%): {recent_avg_growth:.2f}% avg growth rate")

# Trend analysis
print(f"\n�� TREND ANALYSIS:")
if recent_avg_growth > middle_avg_growth > early_avg_growth:
    print("   🔥 ACCELERATING GROWTH: Growth rate is increasing over time")
elif recent_avg_growth < middle_avg_growth < early_avg_growth:
    print("   📉 DECELERATING GROWTH: Growth rate is decreasing over time")
elif abs(recent_avg_growth - early_avg_growth) < 5:
    print("   📊 STABLE GROWTH: Growth rate is relatively consistent")
else:
    print("   🔄 VARIABLE GROWTH: Growth rate shows mixed patterns")

# Peak growth periods
print(f"\n🏔️ PEAK GROWTH PERIODS:")
monthly_growth = df.groupby(df['release_date'].dt.to_period('M')).size()
top_3_months = monthly_growth.nlargest(3)
for i, (month, count) in enumerate(top_3_months.items(), 1):
    print(f"   {i}. {month}: {count} servers")

# Growth momentum analysis
print(f"\n⚡ GROWTH MOMENTUM:")
recent_30_days = df[df['release_date'] >= df['release_date'].max() - pd.Timedelta(days=30)]
recent_60_days = df[df['release_date'] >= df['release_date'].max() - pd.Timedelta(days=60)]

recent_30_count = len(recent_30_days)
recent_60_count = len(recent_60_days)
previous_30_count = recent_60_count - recent_30_count

print(f"   Last 30 days: {recent_30_count} servers")
print(f"   Previous 30 days: {previous_30_count} servers")
if recent_30_count > previous_30_count:
    print(f"   📈 MOMENTUM: Growing (+{recent_30_count - previous_30_count} more servers)")
elif recent_30_count < previous_30_count:
    print(f"   📉 MOMENTUM: Slowing (-{previous_30_count - recent_30_count} fewer servers)")
else:
    print(f"   �� MOMENTUM: Stable (same number of servers)")

# Category growth leaders over time
print(f"\n🏅 CATEGORY GROWTH LEADERS BY PERIOD:")

# Early period
early_categories = early_data['primary_category'].value_counts().head(5)
print(f"   Early Period Top 5:")
for i, (category, count) in enumerate(early_categories.items(), 1):
    print(f"     {i}. {category}: {count} servers")

# Recent period
recent_categories = recent_data['primary_category'].value_counts().head(5)
print(f"   Recent Period Top 5:")
for i, (category, count) in enumerate(recent_categories.items(), 1):
    print(f"     {i}. {category}: {count} servers")

# Growth sustainability analysis
print(f"\n🌱 GROWTH SUSTAINABILITY ANALYSIS:")
total_days = df_sorted['days_since_start'].max()
servers_per_day = total_servers / total_days
print(f"   Current Rate: {servers_per_day:.2f} servers per day")
print(f"   Projected Annual Growth: {servers_per_day * 365:.0f} servers per year")

# Predict future growth (simple linear projection)
if len(growth_rates_30d) > 10:
    recent_growth_trend = np.mean(growth_rates_30d[-10:])  # Last 10 periods
    print(f"   Recent Growth Trend: {recent_growth_trend:.2f}% per 30 days")
    
    # Simple projection for next 6 months
    current_count = total_servers
    projected_6m = current_count * (1 + recent_growth_trend/100) ** 6
    print(f"   Projected 6-month total: {projected_6m:.0f} servers")
```
