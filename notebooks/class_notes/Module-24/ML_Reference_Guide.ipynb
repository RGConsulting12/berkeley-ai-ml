## 13. Best Practices and Tips

### 13.1 General Workflow

1. **Load and Explore Data**
   - Use `df.head()`, `df.info()`, `df.describe()`
   - Check for missing values with `df.isnull().sum()`
   - Visualize distributions and relationships

2. **Clean Data**
   - Handle missing values appropriately
   - Remove or fix outliers
   - Check for duplicate rows

3. **Feature Engineering**
   - Create derived features
   - Encode categorical variables
   - Scale/normalize numeric features
   - Create polynomial features if needed

4. **Split Data**
   - Always use `train_test_split` with a fixed `random_state`
   - Typical split: 70-30 or 80-20
   - Create a baseline model first

5. **Build Models**
   - Start simple (linear regression)
   - Add complexity gradually
   - Use pipelines to avoid data leakage
   - Compare multiple model complexities

6. **Evaluate**
   - Use appropriate metrics (MSE, RMSE, R², etc.)
   - Check both train and test performance
   - Watch for overfitting (train MSE << test MSE)

7. **Iterate**
   - Try regularization if overfitting
   - Use feature selection to reduce complexity
   - Experiment with different preprocessing steps


### 13.2 Common Pitfalls to Avoid

❌ **Don't:**
- Fit scalers on the entire dataset before splitting
- Use test data for any decision-making
- Forget to handle missing values
- Ignore feature scaling for algorithms that need it
- Use polynomial features without regularization
- Forget to set `random_state` for reproducibility

✅ **Do:**
- Always set `random_state` for reproducibility
- Use pipelines to avoid data leakage
- Visualize your data before modeling
- Compare to a baseline model
- Check for overfitting
