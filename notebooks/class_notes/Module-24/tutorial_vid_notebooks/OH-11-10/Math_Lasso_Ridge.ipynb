# Ridge and Lasso Regression — Mathematical Foundations

##1. Why Regularization?
Regularization helps prevent **overfitting** by penalizing large parameter values.  
When a model overfits, it fits both signal and noise — leading to low training error but high testing error.

The idea is to modify the **loss function** to include a *complexity penalty* on the parameters.

Let’s start from ordinary least squares (OLS):

$
\text{Loss}_{OLS} = \frac{1}{n}\sum_{i=1}^n (y_i - \hat{y}_i)^2 = \frac{1}{n}\|y - Xw\|_2^2
$

where  
- $X$ is the feature matrix (size $n \times p$)  
- $y$ is the target vector  
- $w$ is the weight vector (parameters)

The goal is to minimize $\text{Loss}_{OLS}$ with respect to $w$.

---

##2. Ridge Regression (L2 Regularization)

### Idea:
Add a penalty proportional to the **sum of squares of the weights** (the L2 norm).

$
\text{Loss}_{Ridge} = \frac{1}{n}\|y - Xw\|_2^2 + \lambda \|w\|_2^2
$

where  
$
\|w\|_2^2 = \sum_{j=1}^p w_j^2
$
and $\lambda \ge 0$ controls the strength of regularization.

- If $\lambda = 0$, we recover OLS.  
- If $\lambda$ is large, weights are heavily shrunk toward zero.

### Closed-form solution
The beauty of Ridge regression is that it has a closed-form solution:

$
w_{ridge} = (X^TX + \lambda I)^{-1} X^T y
$

This term $\lambda I$ (the **identity matrix scaled by λ**) makes $(X^TX + \lambda I)$ invertible, even when $X^TX$ is ill-conditioned.  
Hence, Ridge also **improves numerical stability**.

---

### Geometric intuition
- Ridge regression constrains coefficients to lie inside a **circle** (L2 ball).  
- The optimization minimizes the sum of squared errors within that circular region.  
- It tends to **shrink** all coefficients smoothly but rarely drives any of them exactly to zero.

---

##3.  Lasso Regression (L1 Regularization)

### Idea:
Add a penalty proportional to the **sum of absolute values of the weights** (the L1 norm).

$
\text{Loss}_{Lasso} = \frac{1}{n}\|y - Xw\|_2^2 + \lambda \|w\|_1
$

where  
$
\|w\|_1 = \sum_{j=1}^p |w_j|
$

- The **L1 penalty** encourages sparsity: it can make some $w_j = 0$ exactly.  
- Hence, Lasso acts as both a **regularizer** and a **feature selector**.

---

### Geometric intuition
- Lasso constrains coefficients inside a **diamond** (L1 ball).  
- The corners of this diamond correspond to some coefficients being exactly zero.  
- When the loss function contour (elliptical) first touches a corner, that weight is set to zero.

---

### Comparison: Ridge vs Lasso

| Aspect | Ridge (L2) | Lasso (L1) |
|:--|:--|:--|
| Penalty | $\lambda \sum w_j^2$ | $\lambda \sum |w_j|$ |
| Constraint shape | Circle (smooth) | Diamond (sharp corners) |
| Effect | Shrinks all coefficients | Forces some to exactly 0 |
| Feature selection |  No |  Yes |
| Closed-form solution |  Yes |  No (requires iterative optimization) |

---

## 4. Unified View: Elastic Net

Sometimes we combine both penalties:

$
\text{Loss}_{ElasticNet} = \frac{1}{n}\|y - Xw\|_2^2 + \lambda_1 \|w\|_1 + \lambda_2 \|w\|_2^2
$

This balances **sparsity (L1)** and **stability (L2)** — useful when you have correlated features.

---

## 5. Effect of λ (Regularization Strength)
- Small $\lambda$ → minimal regularization → may overfit.  
- Large $\lambda$ → strong penalty → may underfit.  
- The optimal $\lambda$ is usually found via **cross-validation**.

---

## 6. Summary

| Concept | Formula | Purpose |
|:--|:--|:--|
| Ridge | $\text{MAE/MSE} + \lambda\sum w_j^2$ | Shrinks weights smoothly |
| Lasso | $\text{MAE/MSE} + \lambda\sum |w_j|$ | Shrinks and zeroes some weights |
| Elastic Net | Combination of both | Balanced shrinkage and selection |

---

**In essence:**
- Ridge makes the model simpler and more stable by shrinking parameters.  
- Lasso simplifies by **zeroing out** unimportant parameters.  
- Both prevent overfitting by limiting model flexibility.
