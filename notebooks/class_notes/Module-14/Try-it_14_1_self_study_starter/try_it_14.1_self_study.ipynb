```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

# Train/validation split (do this ONCE)
X_train, X_val, y_train, y_val = train_test_split(
    X, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

depths = range(1, 16)
train_scores = []
val_scores = []

for d in depths:
    model = DecisionTreeClassifier(max_depth=d, random_state=42)
    model.fit(X_train, y_train)

    train_scores.append(model.score(X_train, y_train))
    val_scores.append(model.score(X_val, y_val))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.plot(depths, train_scores, marker='o', label='Training Accuracy')
plt.plot(depths, val_scores, marker='o', label='Validation Accuracy')

plt.axvline(x=4, color='red', linestyle='--', label='Overfitting Starts')

plt.xlabel("Tree Depth")
plt.ylabel("Accuracy")
plt.title("Decision Tree: Training vs Validation Accuracy")
plt.legend()
plt.grid(True)

plt.show()
```

```python
import numpy as np

children_left = dtree.tree_.children_left
children_right = dtree.tree_.children_right
impurity = dtree.tree_.impurity

n_nodes = dtree.tree_.node_count
node_depth = np.zeros(n_nodes, dtype=int)

stack = [(0, 0)]  # (node_id, depth), start at root

while stack:
    node_id, depth = stack.pop()
    node_depth[node_id] = depth

    left = children_left[node_id]
    right = children_right[node_id]

    if left != -1:
        stack.append((left, depth + 1))
    if right != -1:
        stack.append((right, depth + 1))
```

```python
from collections import defaultdict

depth_impurity = defaultdict(list)

for i in range(n_nodes):
    depth_impurity[node_depth[i]].append(impurity[i])

depths = sorted(depth_impurity.keys())
avg_impurity = [np.mean(depth_impurity[d]) for d in depths]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.plot(depths, avg_impurity, marker='o')

plt.xlabel("Tree Depth")
plt.ylabel("Average Gini Impurity")
plt.title("Average Gini Impurity vs Tree Depth")
plt.grid(True)

plt.show()
```

```python
import numpy as np

tree = dtree.tree_

n_nodes = tree.node_count
impurity = tree.impurity
n_samples = tree.n_node_samples
children_left = tree.children_left
children_right = tree.children_right

impurity_reduction = np.zeros(n_nodes)

for i in range(n_nodes):
    left = children_left[i]
    right = children_right[i]

    # Skip leaf nodes
    if left == -1 or right == -1:
        continue

    w_left = n_samples[left] / n_samples[i]
    w_right = n_samples[right] / n_samples[i]

    impurity_reduction[i] = (
        impurity[i]
        - (w_left * impurity[left] + w_right * impurity[right])
    )
```
