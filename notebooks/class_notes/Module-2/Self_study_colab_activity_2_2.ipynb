### Problem 1: A Uniform Distribution

###  Problem 2: Loop of Samples



Use a `for` loop to generate samples of sizes from 1 to 500. Inside the `for` loop, use the `.rvs` method on `dist1` to return a random sample. Inside the `.rvs` method, set `random_state` equal to 22.  Use `np.mean` to compute the sample mean of each sample size. Append each sample mean to the list `sample_means`.  

```python


sample_means = []

dist1 = uniform(loc = 5, scale = 10)
sample_means = []
for i in range(1, 501):
    sample = dist1.rvs(i, random_state = 22)
    sample_mean = np.mean(sample)
    sample_means.append(sample_mean)


# Answer check
print(type(sample_means))
print(sample_means[:5], '\n', sample_means[-5:])
```

### Problem 5: Repeat with Gaussian Distribution



Now, you are to repeat the above exercise using samples from a Gaussian distribution


Use the function `norm` with argument `loc` equal to five and `scale` equal to ten to define a Gaussian distribution and assign the result to `gauss_dist`.

Use a `for` loop to generate samples of sizes from 30 to 500. Inside the `for` loop, use the `.rvs` method on `gauss_dist` to return a random sample. Inside the `.rvs` method, set `random_state` equal to 22.  Use `np.mean` to compute the sample mean of each sample size. Append each sample mean to the list `sample_means_gauss`.  

Outside the `for` loop, use `np.mean()` to compute the mean and assign the result to the variable `gauss_mean`.

Outside the `for` loop, use `np.std()` to compute the standard deviation and assign the result to the variable `gauss_standard_deviation`.

```python

gauss_dist = norm(loc = 5, scale = 10)
sample_means_gauss = []
for i in range(30, 501):
    sample = gauss_dist.rvs(i, random_state = 22)
    sample_mean = np.mean(sample)
    sample_means_gauss.append(sample_mean)

gauss_mean = np.mean(sample_means_gauss)
gauss_standard_deviation = np.std(sample_means_gauss)


# Answer check
print(type(sample_means_gauss))
print(sample_means_gauss[:5], '\n', sample_means_gauss[-5:])
print(gauss_mean)
print(gauss_standard_deviation)
```

```python
import numpy as np

# Population data
salaries = bball["salary"]

def sample_means(sample_size, n_samples=1000):
    means = []
    for _ in range(n_samples):
        sample = np.random.choice(salaries, size=sample_size, replace=True)
        means.append(sample.mean())
    return np.array(means)

# Example: different sample sizes
means_5   = sample_means(5)
means_30  = sample_means(30)
means_100 = sample_means(100)
```

```python
import matplotlib.pyplot as plt

plt.hist(means_5, bins=30, alpha=0.5, label="n=5")
plt.hist(means_30, bins=30, alpha=0.5, label="n=30")
plt.hist(means_100, bins=30, alpha=0.5, label="n=100")
plt.axvline(salaries.mean(), color="red", linestyle="--", label="Population Mean")
plt.legend()
plt.title("Sampling Distributions of Salary Means")
plt.xlabel("Sample Mean Salary")
plt.ylabel("Frequency")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse

# Extract two numeric columns
X = bball[["salary", "gamesplayed"]].dropna()
x_mean, y_mean = X.mean()
cov = np.cov(X["salary"], X["gamesplayed"])

# Eigen decomposition (for ellipse orientation/scale)
eigvals, eigvecs = np.linalg.eigh(cov)
order = eigvals.argsort()[::-1]
eigvals, eigvecs = eigvals[order], eigvecs[:, order]

theta = np.degrees(np.arctan2(*eigvecs[:,0][::-1]))  # rotation angle

# Function to draw multiple ellipses
def plot_ellipse(mean, eigvals, eigvecs, n_std, **kwargs):
    """Draw an error ellipse with n_std standard deviations."""
    width, height = 2 * n_std * np.sqrt(eigvals)
    ellipse = Ellipse(xy=mean, width=width, height=height,
                      angle=theta, **kwargs)
    return ellipse

# Scatter plot of data
plt.scatter(X["salary"], X["gamesplayed"], alpha=0.3)

# Add 1σ, 2σ, 3σ ellipses
for n_std, color in zip([1, 2, 3], ["red", "green", "blue"]):
    ellipse = plot_ellipse((x_mean, y_mean), eigvals, eigvecs, n_std,
                           edgecolor=color, facecolor="none", lw=2, label=f"{n_std}σ")
    plt.gca().add_patch(ellipse)

plt.xlabel("Salary")
plt.ylabel("Games Played")
plt.title("Error Ellipses (1σ, 2σ, 3σ) for Salary vs Games Played")
plt.legend()
plt.show()
```

```python
import numpy as np

df = bball.copy()
y = df["high_salary"].to_numpy()

# Baseline: global probability
p_global = y.mean()
p_hat_global = np.full_like(y, fill_value=p_global, dtype=float)

# Position-based model
pos_rate = df.groupby("position")["high_salary"].mean()
p_hat_pos = df["position"].map(pos_rate).to_numpy()

# Cross-entropy
eps = 1e-12
def cross_entropy(y_true, y_prob):
    y_prob = np.clip(y_prob, eps, 1 - eps)
    return -np.mean(y_true*np.log(y_prob) + (1-y_true)*np.log(1-y_prob))

print("Cross-entropy (global):", cross_entropy(y, p_hat_global))
print("Cross-entropy (position):", cross_entropy(y, p_hat_pos))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(
    bball_log, 
    vars=['log_salary', 'gamesplayed'], 
    hue='salary_quartile', 
    diag_kind='kde', 
    plot_kws={'alpha': 0.7}
)

plt.suptitle('Pairplot: Log(Salary) vs Games Played by Salary Quartile', y=1.02)
plt.show()
```
