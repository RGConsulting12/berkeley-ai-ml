```python
q1 = nyc_salary_data["base_salary"].quantile(0.25)
q3 = nyc_salary_data["base_salary"].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
salaries_no_outliers = nyc_salary_data.loc[
    (nyc_salary_data["base_salary"] >= lower_bound) &
    (nyc_salary_data["base_salary"] <= upper_bound)
]
```

```python
print("Original size:", len(nyc_salary_data))
print("After removing outliers:", len(salaries_no_outliers))
```

```python
import pandas as pd

# Example: salaries and games played
cov_matrix = nyc_salary_data[["base_salary", "regular_hours"]].cov()
corr_matrix = nyc_salary_data[["base_salary", "regular_hours"]].corr()

print("Covariance Matrix:\n", cov_matrix)
print("\nCorrelation Matrix:\n", corr_matrix)
```

```python
from scipy import stats
import numpy as np

x = nyc_salary_data["base_salary"]
y = nyc_salary_data["regular_hours"]

# Covariance
cov = np.cov(x, y, ddof=1)  # 2×2 matrix
print("Covariance matrix:\n", cov)

# Correlation
corr, p_value = stats.pearsonr(x, y)
print("Correlation:", corr, "p-value:", p_value)
```
